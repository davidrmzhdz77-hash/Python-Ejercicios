```python
#Rango de edad
edad = int(input("Escribe tu edad"))
if edad <= 11 and edad >=0:
    print("Eres un(a) niño/a")
elif edad <= 17 and edad >=12:
    print("Eres adolescente")
elif edad <= 29 and edad >=18:
    print("Eres joven")
elif edad <= 64 and edad >=30:
    print("Eres adulto")
elif edad >=65:
    print("Eres adulto mayor")
else:
    print("Edad negativa. intenta de nuevo")
```

```python
#Número positivo, negativo o cero
numero = int(input("Escribe un número"))
if numero < 0:
    print("El número es negativo")
elif numero >0:
    print("El número es positivo")
else:
    print("El número es cero")
```

```python
#Número par o impar
numero = int(input("Escribe un número"))
if numero % 2 == 0:
    print("El número es par")
else:
    print("El número es impar")
```

```python
#Para subir a una montaña rusa, una persona debe cumplir ambas condiciones: medir al menos 140cm y tener al menos 12 años
altura = int(input("Escribe tu altura en cm"))
edad = int(input("Escribe tu edad"))
if altura >= 140 and edad >= 12:
    print("Puedes subir a la montaña rusa")
else:
    print("No puedes subir a la montaña rusa")
```

```python
import math
a = 7
b = 8
c = 9
semiperimetro = ((a+b+c) / (2))
Area = math.sqrt(semiperimetro*(semiperimetro-a)*(semiperimetro-b)*(semiperimetro-c))
print(f"El área del triángulo es: {Area}")
```
